```python
import sqlite3

conn = sqlite3.connect("shop.db")
cursor = conn.cursor()
print("Database connected sucessfully!")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS customers (
    customer_id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS products (
    product_id INTEGER PRIMARY KEY,
    product_name TEXT,
    price REAL
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    product_id INTEGER,
    quantity INTEGER,
    amount REAL,
    FOREIGN KEY(customer_id) REFERENCES customers(customer_id),
    FOREIGN KEY(product_id) REFERENCES products(product_id)
)
""")

conn.commit()

print("All tables created successfully!")
```

```python
customers = [
    (1, "Manoj", "Hyderabad"),
    (2, "Rahul", "Bangalore"),
    (3, "Suresh", "Chennai"),
    (4, "Priya", "Hyderabad")
]

products = [
    (1, "Laptop", 55000),
    (2, "Smartphone", 25000),
    (3, "Headphones", 3000),
    (4, "Keyboard", 2000)
]

orders = [
    (1, 1, 1, 1, 55000),
    (2, 1, 3, 2, 6000),
    (3, 2, 2, 1, 25000),
    (4, 3, 1, 2, 110000),
    (5, 3, 4, 2, 4000),
    (6, 4, 2, 1, 25000)
]

cursor.executemany(
    "INSERT OR IGNORE INTO customers VALUES (?, ?, ?)",
    customers
)

cursor.executemany(
    "INSERT OR IGNORE INTO products VALUES (?, ?, ?)",
    products
)

cursor.executemany(
    "INSERT OR IGNORE INTO orders VALUES (?, ?, ?, ?, ?)",
    orders
)

conn.commit()

print("Sample data inserted successfully!")
```

```python
print("CUSTOMERS:")
cursor.execute("SELECT * FROM customers")
for row in cursor.fetchall():
    print(row)

print("\nPRODUCTS:")
cursor.execute("SELECT * FROM products")
for row in cursor.fetchall():
    print(row)

print("\nORDERS:")
cursor.execute("SELECT * FROM orders")
for row in cursor.fetchall():
    print(row)
```

```python
from langchain_community.utilities import SQLDatabase

# Connect LangChain to our SQLite database
db = SQLDatabase.from_uri("sqlite:///shop.db")

print("Database dialect:", db.dialect)
print("Available tables:", db.get_usable_table_names())
```

```python
from langchain_community.agent_toolkits import SQLDatabaseToolkit

# Create a toolkit containing database tools
toolkit = SQLDatabaseToolkit(
    db=db,
    llm=llm
)

tools = toolkit.get_tools()

print("Available SQL tools:")
for tool in tools:
    print("-", tool.name)
```

```python
from langchain.agents import create_agent

system_prompt = """
You are a helpful SQL database assistant.

You have access to a shop database.

Follow these rules:
1. First identify the relevant tables.
2. Check the table schema before writing SQL.
3. Generate a valid SQLite SQL query.
4. Check the SQL query before executing it.
5. Only use SELECT queries.
6. Never modify or delete database data.
7. After getting the result, explain the answer clearly.
"""

agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt=system_prompt
)

print("Text-to-SQL agent created successfully!")
```

```python
from langchain.agents import create_agent

system_prompt = """
You are a helpful SQL database assistant.

You have access to a shop database.

Follow these rules:
1. First identify the relevant tables.
2. Check the table schema before writing SQL.
3. Generate a valid SQLite SQL query.
4. Check the SQL query before executing it.
5. Only use SELECT queries.
6. Never modify or delete database data.
7. After getting the result, explain the answer clearly.
"""

agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt=system_prompt
)

print("Text-to-SQL agent recreated successfully!")
```

```python
question = "Who spent the most money?"

result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": question
        }
    ]
})

print(result["messages"][-1].content)
```
